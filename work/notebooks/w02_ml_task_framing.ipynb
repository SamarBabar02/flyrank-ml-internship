## ML Task Type

**Ranking**

I chose **Ranking** because the goal is to prioritize web pages based on their need for content review. The model will rank pages from highest to lowest priority so that SEO specialists or content managers can decide which pages should be reviewed first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or Proxy

The target is to predict which web pages should be prioritized for content review. Since there is no direct "needs update" label in the starter dataset, I will use a proxy based on observed search performance signals, such as declining trends and engagement metrics. The label is therefore based on observed outcomes rather than a manually defined rule.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric

**Precision@50**

I will use Precision@50 because the goal is to rank the most important pages for review. A higher Precision@50 means the model is recommending more useful pages at the top of the list.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Fixed Rule

A fixed rule cannot capture the complex relationships between multiple signals such as impressions, CTR, trends, and engagement. Different pages may require different decisions even when one metric looks similar. Machine Learning can learn these patterns from data and provide better ranking and prioritization than a single if-statement.
